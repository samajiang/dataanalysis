```python
#数据过滤
#配置项，展示所有数据
# pd.set_option('display.max_rows',None)

df4 = pd.read_csv("./resource/sales.csv")
df4
#获取销售数量=>10的订单数据
df4[df4['销售数量']>=10]
#获取产品类别是图书或电子产品的订单数据
df4[(df4['产品类别']=='图书')|(df4['产品类别']=='电子产品')]
df4[df4['产品类别'].isin(['图书','电子产品'])]
#获取单价在100-200之间的订单数据
df4[(df4['单价']>=100)&(df4['单价']<=200)]
df4[df4['单价'].between(100,200)]#默认包含边界值between(100,200, inclusive='left')表示包含左边界，exclusive='right'表示包含右边界
#获取销售额>=8且单价>=100的订单数据
df4[(df4['销售数量']>=8)&(df4['单价']>=100)]
#获取产品类别是服装\图书且支付方式为支付宝\微信的订单数据
df4[(df4['产品类别'].isin(['服装','图书']))&(df4['支付方式'].isin(['支付宝','微信支付']))]



```

```python
#数据清洗
import pandas as pd

#限制显示行数
pd.set_option('display.min_rows',30)
#读取数据
df5 = pd.read_csv("./resource/sales.csv")
df5.head(15)

#缺失值处理

#1查看缺失值
df5.isnull()
#2处理缺失值
#删除包含缺失值的行
df5.dropna()
#删除包含缺失值的列
df5.dropna(axis=1)
#填充缺失值
df5.fillna("想填的值")
#用前一行的值填充缺失值
df5.ffill()
#用后一行的值填充缺失值
df5.bfill

#重复值处理

#1查看重复值
df5.duplicated()#所有的列的数据都重复才会是false
df5.duplicated(subset=['订单号'])#指定列的数据重复才会是false

#2处理重复值

#删除重复值
df5.drop_duplicates(subset=['订单号'],keep='first')#默认保留第一个，keep='last'保留最后一个，keep=False不保留任何

#3异常值处理

#1查看异常值
df5[df5['单价']<0]#单价小于0的异常值
#2处理异常值
#删除异常值
df5.drop(df5[df5['单价']<0].index)
#替换异常值
df5['单价'] = df5['单价'].abs()#取绝对值
df5['单价'].replace(-29,29)

#数据格式处理
df5["订单日期"]=df5["订单日期"].str.replace("/","-")#str是拿到这个字符串的编辑权限
df5



```
