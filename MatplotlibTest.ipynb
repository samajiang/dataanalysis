```python
import random

import matplotlib.pyplot as plt
from matplotlib.lines import lineStyles
from pyparsing import alphas

#这个matplotlib库不支持显示中文，需要设置字体
plt.rcParams['font.sans-serif'] = ['SimHei']#设置字体为黑体，解决中文显示问题
#xy的数据个数要一致
x = [i for i in range(1,25)]#生成1到25的列表
y = [random.randint(10,25) for i in x]#生成10到25的随机数列表
yjj = [random.randint(13,18) for i in x]
y_ticks = [i for i in range(5,24)]
plt.figure(figsize=(10,5))#创建一个画布，设置大小为10*5宽，高（英寸）
plt.plot(x,y,label='bj')#plot折线图(如果没有画布，会自动创建一个)
plt.plot(x,yjj,label='sh')
#设置曲线图详细信息
plt.title("折线图示例",fontsize=16)#设置标题
plt.xlabel("时间",fontsize=14)#设置X轴标签
plt.ylabel("温度",fontsize=14)#设置Y轴标签
plt.xticks(x)#把x轴刻度全部展示
plt.yticks(y_ticks)#把y轴刻度全部展示
plt.grid(linestyle='--',alpha=0.1)#设置网格,lineStyle设置网格线样式,alpha设置网格线透明度
plt.legend(loc='lower right')#设置图例,loc设置图例位置
plt.show()#展示图表
```
