```python
import random

import matplotlib.pyplot as plt
from fontTools.diff import color
from matplotlib.lines import lineStyles
from matplotlib.patches import bbox_artist
from pyparsing import alphas

#这个matplotlib库不支持显示中文，需要设置字体
plt.rcParams['font.sans-serif'] = ['SimHei']#设置字体为黑体，解决中文显示问题
#xy的数据个数要一致
x = [i for i in range(1,25)]#生成1到25的列表
y = [random.randint(10,25) for i in x]#生成10到25的随机数列表
yjj = [random.randint(13,18) for i in x]
y_ticks = [i for i in range(5,24)]
plt.figure(figsize=(10,5))#创建一个画布，设置大小为10*5宽，高（英寸）
plt.plot(x,y,label='bj')#plot折线图(如果没有画布，会自动创建一个)
plt.plot(x,yjj,label='sh')
#设置曲线图详细信息
plt.title("折线图示例",fontsize=16)#设置标题
plt.xlabel("时间",fontsize=14)#设置X轴标签
plt.ylabel("温度",fontsize=14)#设置Y轴标签
plt.xticks(x)#把x轴刻度全部展示
plt.yticks(y_ticks)#把y轴刻度全部展示
plt.grid(linestyle='--',alpha=0.1)#设置网格,lineStyle设置网格线样式,alpha设置网格线透明度
plt.legend(loc='lower right')#设置图例,loc设置图例位置
plt.show()#展示图表
```

```python
#创建子图
import matplotlib.pyplot as plt
#这个matplotlib库不支持显示中文，需要设置字体
plt.rcParams['font.sans-serif'] = ['SimHei']#设置字体为黑体，解决中文显示问题

#创建子图
#fig：创建的图形对象，axes：子图数组
fig,axes = plt.subplots(1,2,figsize=(10,5))#创建一个1行2列的子图，设置大小为10*5宽，高（英寸）
#创建子图1(柱状图)
axes1:plt.Axes = axes[0]
#数据
x = ['中国','美国','英国','日本','韩国']
y = [100,80,60,40,20]

#设置子图1的参数
axes1.set_title("jjww",fontsize=16)#标题
axes1.set_xlabel("国家",fontsize=14)#X轴标签
axes1.set_ylabel("石油(亿吨)",fontsize=14)#Y轴标签
axes1.bar(x,y,color='r',width=0.5)#柱状图(如果没有画布，会自动创建一个)，color设置颜色，width设置宽度

#创建子图2(饼状图)
axes2:plt.Axes = axes[1]#子图2
#数据
labels = ['中国','美国','英国','日本','韩国']
data = [14.5,13.5,12.5,11.5,10.5]
axes2.pie(data,labels=labels,autopct='%1.2f%%')#饼状图(如果没有画布，会自动创建一个)，labels设置标签，autopct设置百分比格式,f表示浮点数，.2表示小数点后保留两位，%表示百分比，%%表示百分号，1这一位表示带上小数点一共多少位，如果是50那就表示数据一共有50位，不够的会在前边加0
axes2.set_title("人口占比")#标题
axes2.legend(loc = 'lower center',ncol = 5,bbox_to_anchor=(0.5,-0.2))#图例,loc设置图例位置，ncol设置图例列数显示成多少列,bbox_to_anchor设置图例位置,0.5表示水平居中，-0.2表示垂直方向向下移动0.2
#保存图表图片
plt.savefig('./resource/chart.png')
```
